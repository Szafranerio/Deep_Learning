```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

Th e cvCanny() function expects an input image, which must be grayscale, and an output image, which must also be grayscale (but which will actually be a Boolean image). The next two arguments are the low and high thresholds, and the last argument is another aperture. 

```python
tennis = cv2.imread('../../DATA/tennis.jpg', 0)
plt.imshow(tennis, cmap='gray')
```

```python
edges = cv2.Canny(tennis, 20, 30)
plt.imshow(edges, cmap='gray')
```

```python
edges = cv2.Canny(tennis, 30, 20)
plt.imshow(edges, cmap='gray')
```

```python
edges = cv2.Canny(tennis, 200, 300)
plt.imshow(edges, cmap='gray')
```

```python
edges = cv2.Canny(tennis, 300, 200)
plt.imshow(edges, cmap='gray')
```
