```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_pic():
    img = cv2.imread('../../DATA/sammy.jpg').astype(np.float32) / 255
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return img
```

```python
def display_pic(img):
    fig = plt.figure(figsize=(12,10))
    ax = fig.add_subplot(111)
    ax.imshow(img)
    
```

```python
sammy = load_pic()
display_pic(sammy)
```

The median blur filter replaces each pixel by the median or “middle” pixel (as opposed to the mean pixel) value in a square neighborhood around the center pixel. Median filter will work on single-channel or three-channel or four-channel 8-bit images, but it cannot be done in place. Results of median filtering are shown in Figure 5-2. Simple blurring by averaging can be sensitive to noisy images, especially images with large isolated outlier points (sometimes called “shot noise”). Large differences in even a small number of points can cause a noticeable movement in the average value. Median filtering is able to ignore the outliers by selecting the middle points.

Bilateral filtering takes two parameters. Th e fi rst is the width of the Gaussian kernel used in the spatial domain, which is analogous to the sigma parameters in the Gaussian filter. Th e second is the width of the Gaussian kernel in the color domain. Th e larger this second parameter is, the broader is the range of intensities (or colors) that will be included in the smoothing (and thus the more extreme a discontinuity must be in order to be preserved).

```python
#Bilateral filtering
animal_bilateral = cv2.bilateralFilter(animal, 150, 75, 75)
display_pic(animal_bilateral) 
```
