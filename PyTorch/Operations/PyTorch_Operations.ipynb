```python
import torch
import numpy as np
```

```python
#Indexing and slicing
x = torch.arange(6).reshape(3,2)
x
```

```python
print(x[1,0])
print(x[0,1])
print(x[2,1])
print(x[1,1])
```

```python
print(x[:,1])
print(x[:,0])
```

```python
print(x[0,:])
print(x[1,:])
print(x[2,:])
```

```python
#View and reshape
x = torch.arange(10)
x.view(2,5)
```

```python
#View and reshape
x = torch.arange(6).reshape(2,3)
print(x)
x = x.t()
print(x)
```
