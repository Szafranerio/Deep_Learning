```python
import torch
```

```python
#Tensor obejct with auto grad
x = torch.tensor(2.0, requires_grad=True)

y = 2*x**4 + x**3 + 3*x**2 + 5*x + 1
print(y) 
```

```python
y.backward()
```

```python
x.grad
```

```python
x = torch.tensor([[1.,2., 3.], [3.,2.,1.]], requires_grad=True)
print(x)
```

```python
y = 3*x + 2
print(y)
```

```python
z = 2*y**2
print(z)
```

```python
out = z.mean()
print(out)
```

```python
out.backward()
print(x.grad)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1️⃣ Define the x values
x = np.linspace(-3, 3, 100)

# 2️⃣ Define the function and its derivative
y = x**2
dy_dx = 2*x

# 3️⃣ Plot them
plt.figure(figsize=(8, 6))

# Original function f(x)
plt.plot(x, y, label='f(x) = x²', color='blue', linewidth=2)

# Derivative f'(x)
plt.plot(x, dy_dx, label="f'(x) = 2x", color='red', linestyle='--', linewidth=2)

# Add labels and legend
plt.title("Function and its Derivative")
plt.xlabel("x")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.show()
```
