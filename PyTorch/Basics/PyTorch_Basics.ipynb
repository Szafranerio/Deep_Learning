```python
import torch
import numpy as np
```

```python
# Scalar tensor
x = torch.tensor(5)
print(x)
print(x.size)
print(x.shape)
print(x.ndim)
```

```python
# Vector tensor
x = torch.tensor([1,2,3,4])
print(x)
print(x.size)
print(x.shape)
print(x.ndim)
```

```python
# Matrix tensor
x = torch.tensor([[1,2,3], [4,5,6]])
print(x)
print(x.size)
print(x.shape)
print(x.ndim)
```

```python
# 3D tenosr
x = torch.rand(2, 3, 4)
print(x)
print(x.shape)
print(x.ndim)
```

```python
x = np.array([1,2,3,4,5])
x
```

```python
x = torch.from_numpy(x)
x
```

```python
#2D
x = np.arange(0,12)
x = x.reshape(4,3)
x
```

```python
x = torch.from_numpy(x)
x
```

```python
#Direct link
x = np.arange(1,12)
print(x)
tensor = torch.from_numpy(x)
print(tensor)

x[0] = 99
print(x)
print(tensor)
```

```python
#To avoid direct link
x = np.arange(1,12)
print(x)
tensor = torch.tensor(x)
print(tensor)

x[0] = 99
print(x)
print(tensor)
```

```python
#Create tensors in PyTorch
x = np.array([1,2,3])
print(f'It will create tensor from numpy and keep data type: {torch.tensor(x)}')
print(f'It will create tensor from numpy but will change data type to float: {torch.Tensor(x)}')
```

```python
#Manual seed to keep same values always
torch.manual_seed(42) #alwyas use it in the beggining if you know you want to keep same vlaues
print(torch.rand(2,3))
torch.manual_seed(42)
print(torch.rand(2,3))
torch.manual_seed(42)
print(torch.rand(2,3))
```

```python
#Without manual_seed it will be different
print(torch.rand(2,3))
print(torch.rand(2,3))
print(torch.rand(2,3))
```
