Gamma correction to make img brighter or darker

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_pic():
    img = cv2.imread('../DATA/bricks.jpg').astype(np.float32) / 255
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return img
```

```python
def display_pic(img):
    fig = plt.figure(figsize=(12,10))
    ax = fig.add_subplot(111)
    ax.imshow(img) 
```

```python
i = load_pic()
display_pic(i)
```

```python
#Gamma correction, over 1 makes it darker
gamma = 2
res = np.power(i,gamma)
display_pic(res)
```

```python
#Less value make it brighter
gamma = 1/22
res = np.power(i,gamma)
display_pic(res)
```

```python
#Bluring (low pass filter)
img = load_pic()
font = cv2.FONT_HERSHEY_COMPLEX
cv2.putText(img, text='bricks', org=(10,600), fontFace = font, fontScale=10, color=(255, 0, 0), thickness=4)
display_pic(img)
```

```python
kernel = np.ones(shape=(5,5), dtype = np.float32)/25
dst = cv2.filter2D(img, -1, kernel)
display_pic(dst)
```

```python
#Smooth
img = load_pic()
font = cv2.FONT_HERSHEY_COMPLEX
cv2.putText(img, text='bricks', org=(10,600), fontFace = font, fontScale=10, color=(255, 0, 0), thickness=4)
print('reset')
```

```python
blurred = cv2.blur(img, ksize=(50,50))
display_pic(blurred)
```

```python
#Gaussian
img = load_pic()
font = cv2.FONT_HERSHEY_COMPLEX
cv2.putText(img, text='bricks', org=(10,600), fontFace = font, fontScale=10, color=(255, 0, 0), thickness=4)
print('reset')
```

```python
blurred_gausian = cv2.GaussianBlur(img,(5,5),10)
display_pic(blurred_gausian)
```
